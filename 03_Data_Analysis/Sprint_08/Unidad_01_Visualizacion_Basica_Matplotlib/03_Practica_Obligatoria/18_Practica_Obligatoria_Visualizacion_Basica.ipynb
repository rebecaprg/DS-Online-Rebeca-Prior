Las figuras son:

```python
#Las ciudades más pobladas, con los numeros de las poblaciones 
#(tendrás que usar texto y recordar que para centrarlos se necesita poner el argumento ha a "center"). 
#(Colores: azul, gris, aceituna, marrón, con trasparencia)
```

```python
#GRÁFICA 1
plt.bar(
    df_casas2["city"],
    df_casas2["population_total"],
    color=["cornflowerblue", "lightgray", "lightcoral", "darkkhaki", "lightpink"]
)

plt.ylabel("Habitantes (miles)")
plt.xlabel("Ciudades")
plt.title("Ciudades más pobladas de California")
```

```python
#GRÁFICA 2
plt.hist(
    df_casas2["area_total_sq_mi"],
	bins = 50,
	alpha = 0.7,
	color = "steelblue"
	)

font_dict = {
	"size": 18,
	"family": "Bradley Hand ITC",
}

plt.xlabel("Superficie (km2)",fontdict=font_dict)
plt.ylabel("Número de Ciudades",fontdict=font_dict)

plt.legend();
```

```python
#El scatter de área de agua de un ciudad con su poblacion.
```

```python
plt.scatter(
    df_casas2["area_water_sq_mi"],
    df_casas2["population_total"],
    alpha=0.6,
    color="green"
)

plt.xlabel("Área de agua en la ciudad (km²)",size = 8)
plt.ylabel("Población (miles habitantes)",size = 8)
plt.title("Población por kilómetro cuadrado de agua", size = 8)

```

```python
#GRAFICA 4
plt.boxplot(df_casas2["elevation_m"], whis = 1.5, vert = False)
plt.xlabel("Altitud(m)", size = 8)
plt.title("Diagrama de caja de la Altitud de las ciudades de California", size = 8)
```

```python
fig, axs = plt.subplots(2, 2, figsize=(12, 10))  
axs[0,0].bar(
    df_casas2["city"],
    df_casas2["population_total"],
    color=["cornflowerblue", "lightgray", "lightcoral", "darkkhaki", "lightpink"]
)

plt.ylabel("Habitantes (miles)")
plt.xlabel("Ciudades")
plt.title("Ciudades más pobladas de California")

#FIGURA 2
axs[0,1].hist(
    df_casas2["area_total_sq_mi"],
	bins = 50,
	alpha = 0.7,
	color = "steelblue"
	)
plt.xlabel("Superficie (km2)",fontdict=font_dict)
plt.ylabel("Número de Ciudades",fontdict=font_dict)
font_dict = {
	"size": 18,
	"family": "Bradley Hand ITC",
}


plt.legend();

#FIGURA 3

axs[1,0].scatter(
    df_casas2["area_water_sq_mi"],
    df_casas2["population_total"],
    alpha=0.6,
    color="green"
)

plt.xlabel("Área de agua en la ciudad (km²)",size = 8)
plt.ylabel("Población (miles habitantes)",size = 8)
plt.title("Población por kilómetro cuadrado de agua", size = 8)

#FIGURA 4

axs[1,1].boxplot(df_casas2["elevation_m"], whis = 1.5, vert = False)
plt.xlabel("Altitud(m)", size = 8)
plt.title("Diagrama de caja de la Altitud de las ciudades de California", size = 8)

plt.savefig("C:\Users\rebec\OneDrive\Desktop\bootcamp-ds\DS-Online-Rebeca-Prior/figura.png", dpi=300)
```
